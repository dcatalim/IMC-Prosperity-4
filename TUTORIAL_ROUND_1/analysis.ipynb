```python
# %pip install pandas matplotlib

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("prices_round_0_day_-2.csv", delimiter=";")

df.head()
```

```python
prices = df.pivot(index="timestamp", columns="product", values="mid_price")

print("="*5 + "EMERALDS" + "="*5)
print(prices["EMERALDS"].describe())

print("="*5 + "TOMATOES" + "="*5)
print(prices["TOMATOES"].describe())
```

```python
emeralds = df[df["product"] == "EMERALDS"]["mid_price"]

ax = emeralds.plot(
    x="timestamp",
    title="EMERALDS price",
    ylim=(9990, 10010)

)
ax.ticklabel_format(axis="x", style="plain", useOffset=False)  # optional: disable +1e4 offset
ax.ticklabel_format(axis="y", style="plain", useOffset=False)  # optional: disable +1e4 offset
```

```python
tomatoes = df[df["product"] == "TOMATOES"]["mid_price"]

ax = tomatoes.plot(
    x="timestamp",
    title="TOMATOES price",
)
ax.ticklabel_format(axis="x", style="plain", useOffset=False)  # optional: disable +1e4 offset
```
